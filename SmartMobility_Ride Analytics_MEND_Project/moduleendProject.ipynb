5. Probability

Basic Probability
Formula
P(A)=
Number of favorable outcomes/ Total outcomes


```python
prob_economy = len(df[df['Ride_Category']=="Economy"])/len(df)

print(prob_economy)
```

Conditional Probability
Formula
P(A|B)={P(AintersetionB)}/{P(B)}

```python
peak_economy = len(df[(df['Ride_Category']=="Economy") &
                      (df['Ride_Time']=="Peak")])

peak = len(df[df['Ride_Time']=="Peak"])

conditional_prob = peak_economy/peak

print(conditional_prob)
```

Bayes Theorem
Formula
P(A|B)={P(B|A)P(A)}{P(B)}

```python
P_A = prob_economy
P_B_given_A = conditional_prob
P_B = peak/len(df)

bayes = (P_B_given_A * P_A)/P_B

print(bayes)
```

7. Outlier Detection
Quartiles & IQR
Formula
IQR=Q3−Q1

```python
Q1 = df['Fare_Amount'].quantile(0.25)
Q3 = df['Fare_Amount'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df['Fare_Amount'] < lower) |
              (df['Fare_Amount'] > upper)]

print(outliers)
```

Z-score
Formula
z={x-mu}/{sigma}

```python
df['zscore'] = zscore(df['Fare_Amount'])

outliers_z = df[np.abs(df['zscore']) > 3]

print(outliers_z)
```

8. Inferential Statistics
Formula
CI=xˉ±z(s/root of n)

```python
mean = df['Fare_Amount'].mean()

std = df['Fare_Amount'].std()

n = len(df)

margin = 1.96*(std/np.sqrt(n))

lower = mean - margin
upper = mean + margin

print(lower, upper)
```

Bootstrapping
Explanation

Bootstrapping creates many samples and estimates confidence intervals.

```python
boot_means = []

for i in range(1000):
    sample = df['Fare_Amount'].sample(
        frac=1,
        replace=True
    )

    boot_means.append(sample.mean())

print(np.percentile(boot_means,[2.5,97.5]))
```

9. Hypothesis Testing

Q1 Distance vs Fare

```python
r,p = stats.pearsonr(
    df['Trip_Distance'],
    df['Fare_Amount']
)

print(r,p)
```

Decision

```python
if p < 0.05:
    print("Significant Relationship")
else:
    print("No Significant Relationship")
```

Q2 Premium vs Economy (T-Test)

```python
premium = df[df['Ride_Category']=="Premium"]['Fare_Amount']

economy = df[df['Ride_Category']=="Economy"]['Fare_Amount']

t,p = stats.ttest_ind(
    premium,
    economy
)

print(t,p)
```

Hypothesis

H₀: Mean Premium Fare = Mean Economy Fare

H₁: Mean Premium Fare ≠ Mean Economy Fare

Q3 Chi-Square Test

```python
economy = df[df['Ride_Category']=="Economy"]['Fare_Amount']

premium = df[df['Ride_Category']=="Premium"]['Fare_Amount']

shared = df[df['Ride_Category']=="Shared"]['Fare_Amount']

f,p = stats.f_oneway(
    economy,
    premium,
    shared
)

print(f,p)
```

Q4 ANOVA

```python
economy = df[df['Ride_Category']=="Economy"]['Fare_Amount']

premium = df[df['Ride_Category']=="Premium"]['Fare_Amount']

shared = df[df['Ride_Category']=="Shared"]['Fare_Amount']

f,p = stats.f_oneway(
    economy,
    premium,
    shared
)

print(f,p)
```

Question 2: Premium vs Economy Fare
Descriptive Statistics

```python
premium = df[df['Ride_Category']=="Premium"]['Fare_Amount']

economy = df[df['Ride_Category']=="Economy"]['Fare_Amount']

print("Premium Mean:", premium.mean())
print("Economy Mean:", economy.mean())
```

T-Test

```python
from scipy.stats import ttest_ind

t_stat,p_value = ttest_ind(
    premium,
    economy,
    equal_var=False
)

print("T-statistic:",t_stat)
print("P-value:",p_value)
```

Decision

```python
if p_value < 0.05:
    print("Significant Difference")
else:
    print("No Significant Difference")
```

Question: Are Premium rides significantly more expensive than Economy rides?

Answer:

No. The t-test produced a p-value of 0.6855, which is greater than 0.05. Therefore, there is no statistically significant difference between Premium and Economy ride fares.

Question 4: Peak vs Non-Peak Revenue

Question 5: Outlier Detection

Quartiles

```python
Q1 = df['Fare_Amount'].quantile(0.25)

Q3 = df['Fare_Amount'].quantile(0.75)

IQR = Q3-Q1

print("Q1 =",Q1)
print("Q3 =",Q3)
print("IQR =",IQR)
```

```python
peak = df[df['Ride_Time']=="Peak"]['Fare_Amount']

non_peak = df[df['Ride_Time']=="Non-Peak"]['Fare_Amount']

from scipy.stats import ttest_ind

t,p = ttest_ind(
    peak,
    non_peak,
    equal_var=False
)

print(t,p)
```

Question 8: Confidence Interval

```python
import numpy as np

mean = df['Fare_Amount'].mean()

std = df['Fare_Amount'].std()

n = len(df)

margin = 1.96*(std/np.sqrt(n))

lower = mean-margin

upper = mean+margin

print(lower)
print(upper)
```

Bootstrapping

```python
boot_means = []

for i in range(1000):

    sample = df['Fare_Amount'].sample(
        frac=1,
        replace=True
    )

    boot_means.append(
        sample.mean()
    )

print(
    np.percentile(
        boot_means,
        [2.5,97.5]
    )
)
```

The bootstrap confidence interval is very close to the traditional confidence interval, indicating that the estimate of the average fare is stable and reliable.
