```python
# Task 3
# Function to clean skills
def clean_skills(text):
    text = text.lower()
    text = text.replace(",", " ")
    text = " ".join(text.split())
    return text

job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer","ML Engineer"],
    "Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
    "Skills": [
        "Excel SQL Python",
        "Python ML Statistics",
        "Excel SQL PowerBI",
        "Python SQL Spark",
        "Python ML DeepLearning"
    ]
}
# Apply function to all skills
for i in range(len(job_data["Skills"])):
    job_data["Skills"][i] = clean_skills(job_data["Skills"][i])

# Print result
#print(job_data)
for i in range(len(job_data["Job_ID"])):
    print('ID:', job_data["Job_ID"][i], 
          '| Role:', job_data['Role'][i], 
          '| Company:', job_data['Company'][i], 
          '| Skills:', job_data['Skills'][i])
#lower() → makes text small
#replace(",", " ") → removes commas
#split() + join() → removes extra spaces
#Loop → applies function to all skills
```

```python
# Step 5: Count skills
job_data = {
    "Job_ID": [1, 2, 3],
    "Role": ["Data Analyst", "Data Scientist", "Data Engineer"],
    "Company": ["ABC Corp", "XYZ Ltd", "TechSoft"],
    "Skills": ["Excel SQL Python", "Python ML", "Python SQL"]
}

all_skills = []
for s in job_data["Skills"]:
    all_skills += s.split()

skill_count = {}
for skill in all_skills:
    skill_count[skill] = skill_count.get(skill, 0) + 1

print("Skill Count:", skill_count)
```

```python
#Task 5
job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "ML Engineer"],
    "Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
    "Skills": ["Excel SQL Python", "Python ML Statistics", "Excel SQL PowerBI", "Python SQL Spark", "Python ML DeepLearning"]
}


for i in range(len(job_data["Job_ID"])):
    print('ID:', job_data["Job_ID"][i], 
          '| Role:', job_data['Role'][i], 
          '| Company:', job_data['Company'][i], 
          '| Skills:', job_data['Skills'][i])
# Total number of jobs
total_jobs = len(job_data["Job_ID"])

# Collect all skills
all_skills = []
for skill_text in job_data["Skills"]:
    words = skill_text.lower().split()
    for w in words:
        all_skills.append(w)

# Unique skills
unique_skills = set(all_skills)

# Jobs mentioning python
python_jobs = []
for i in range(len(job_data["Role"])):
    if "python" in job_data["Skills"][i].lower():
        python_jobs.append(job_data["Role"][i])

# Print results
print("\nBasic Statistics:\n")
print("Total Jobs:", total_jobs)
print("Unique skills:", len(unique_skills))
print("Jobs mentioning Python:", python_jobs)
    
```
