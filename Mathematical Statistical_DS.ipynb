```python
import numpy as np
```

```python
# Vectors
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])
```

```python
# Matrices
A = np.array([[2, 1, 3],
              [1, 0, 2],
              [4, 1, 1]])
```

```python
B = np.array([[1, 2, 0],
              [3, 1, 4],
              [2, 5, 1]])
```

```python
# Vector addition
print("Addition:")
print(a + b)
```

```python
# Vector subtraction
print("Subtraction:")
print(a - b)
```

```python
# Dot product
print("Dot Product:")
print(np.dot(a, b))
```

```python
# Matrix addition
print("Matrix Addition:")
print(A + B)
```

```python
# Matrix multiplication
print("Matrix Multiplication:")
print(np.dot(A, B))
```

```python
# Determinant
d = np.linalg.det(A)
print("Determinant:")
print(d)
```

```python
# Inverse
if d != 0:
    print("Inverse:")
    print(np.linalg.inv(A))
```

```python
# Eigenvalues and Eigenvectors
value, vector = np.linalg.eig(A)

print("Eigenvalues:")
print(value)

print("Eigenvectors:")
print(vector)
```

```python
'''#-------------------------------------------
a + b → adds two vectors.
a - b → subtracts two vectors.
np.dot(a, b) → finds dot product.
A + B → adds two matrices.
np.dot(A, B) → multiplies two matrices.
np.linalg.det(A) → finds determinant.
np.linalg.inv(A) → finds inverse.
np.linalg.eig(A) → finds eigenvalues and eigenvectors.
------------------------------------------------------'''
```

```python
# -------------------------
# 3. Bayes Theorem
# -------------------------

P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_A) / P_B

print("\nBayes Theorem Result")
print("P(A|B) =", P_A_given_B)
```

```python
# -------------------------
# 4. Binomial Distribution
# -------------------------

n_trials = 20
success_prob = 0.65

binomial_result = binom.pmf(13, n_trials, success_prob)

print("\nBinomial Distribution")
print("Probability of exactly 13 successes:", binomial_result)

```

```python
#Task-3

import numpy as np
from statistics import mode
from scipy.stats import skew, kurtosis
```

```python
# Dataset
sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
              190, 200, 210, 195, 185, 175, 165, 155, 145, 135]
```

```python
# Mean
print("Mean:", np.mean(sales_data))

# Median
print("Median:", np.median(sales_data))

# Mode
print("Mode:", mode(sales_data))

# Range
print("Range:", max(sales_data) - min(sales_data))

# Variance
print("Variance:", np.var(sales_data))

# Standard Deviation
print("Standard Deviation:", np.std(sales_data))

# Skewness
s = skew(sales_data)
print("Skewness:", s)

# Kurtosis
k = kurtosis(sales_data)
print("Kurtosis:", k)

# Interpretation
if s > 0:
    print("Dataset is Positively Skewed")
elif s < 0:
    print("Dataset is Negatively Skewed")
else:
    print("Dataset is Symmetric")

```

```python
"""
Interpretation
If Skewness = 0, the data is symmetric.
If Skewness > 0, the data is positively skewed (tail on the right).
If Skewness < 0, the data is negatively skewed (tail on the left).

For this dataset, the skewness is very close to 0, so the dataset is approximately symmetric."""
```
