```python
import numpy as np
```

```python
#1.Create a 1D Numpy Array

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("week1 Temperatures:")
print(temperatures_w1)
```

```python
#2 Inspecting and properties
print("\nShape of the array:")
print(temperatures_w1.shape)

print("\nData Type:")
print(temperatures_w1.dtype)

print("\nNumber of elements:")
print(temperatures_w1.size)
```

```python
#Array Operations
#celsius to fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32
print("\nTemperatures in Fahrenheit:")
print(fahrenheit)

print("\nMaximum Temperature:")
print(np.max(temperatures_w1))

print("\nMinimum Temperature:")
print(np.min(temperatures_w1))

print("mean Temperature:")
print(np.mean(temperatures_w1))
```

```python
#4.Array slicing and indexing
print("\nFirst three temperatures:")
print(temperatures_w1[:3])

print("\nweekend temperatures:")
print(temperatures_w1[5:])

print("\nmiddle three days temperatures:")
print(temperatures_w1[2:5])
```

```python
#create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [21.0, 24.5, 19.8, 22.0, 25.0, 23.5, 20.5],  # Week 2
    [23.0, 26.0, 21.5, 24.0, 27.0, 25.5, 22.0]   # Week 3
])

print ("\n2D Array of Temperatures:")
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array
print("\nShape of the 2D array:")
print(temperatures.shape)
print("\ndata type of the 2D array:")
print(temperatures.dtype)
print ("total number of elements in the 2D array:")
print(temperatures.size)
print("\nTemperatures for Week 1:")
print(temperatures[0])
print("\nTemperatures for Week 2:")
print(temperatures[1])
print("\nweekend temperatures for all  weeks:")
print(temperatures[:, -2:])


```

```python
import pandas as pd
```

```python
# Create marks series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

# Print series
print("Student Marks")
print(marks)
```

```python
#2.Indexing and slicing
print("\n1st Rank Mark")
print(marks["Rank1"])
print("\nTop 3 Ranks")
print(marks.loc[["Rank1", "Rank2", "Rank3"]])

print("\n3rd Rank Mark")
print(marks.iloc[2])

print("\nMarks Greater Than 90")
print(marks[marks > 90])
```

```python
# 3. Manipulating Series

marks["Rank1"] = 100
print("\nAfter Changing Rank1 Mark")
print(marks)

marks = marks.drop("Rank5")
print("\nAfter Dropping Rank5")
print(marks)

cgpa = marks / 10
print("\nCGPA")
print(cgpa)
```

```python
# Create dictionary data
data = {
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "ProductCategory": [
        "Electronics",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Furniture",
        "Electronics"
    ],

    "Region": [
        "North",
        "South",
        "North",
        "East",
        "West",
        "North",
        "East",
        "West",
        "South",
        "North"
    ],

    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Create DataFrame  
transactions = pd.DataFrame(data)
print("Transactions DataFrame")
print(transactions)
```

```python
# 2. Data Exploration
# First 5 rows
print("\nHead")
print(transactions.head())
# Last 5 rows
print("\nTail")
print(transactions.tail())
# Shape of DataFrame
print("\nShape")
print(transactions.shape)
# Column names
print("\nColumn Names")
print(transactions.columns)
# Data types
print("\nData Types")
print(transactions.dtypes)
# Show only ProductCategory and Amount
print("\nProductCategory and Amount")
print(transactions[["ProductCategory", "Amount"]])
# Last 3 columns using iloc
print("\nLast 3 Columns")
print(transactions.iloc[:, -3:])

# Filter North region and Amount > 200
print("\nNorth Region and Amount Greater Than 200")
print(transactions[
    (transactions["Region"] == "North") &
    (transactions["Amount"] > 200)
])


# Count ProductCategory values
print("\nProduct Category Count")
print(transactions["ProductCategory"].value_counts())


# Unique Region values
print("\nUnique Regions")
print(transactions["Region"].unique())


# Mean amount by Region
print("\nMean Amount By Region")
print(transactions.groupby("Region")["Amount"].mean())

```

```python
# 3. Manipulating DataFrame
# Change Amount for TransactionID 102
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

print("\nAfter Changing Amount")
print(transactions)


# Add Discount column
# 10% discount
transactions["Discount"] = transactions["Amount"] * 0.10

print("\nAfter Adding Discount")
print(transactions)


# Remove row with TransactionID 109
transactions = transactions[transactions["TransactionID"] != 109]

print("\nAfter Removing TransactionID 109")
print(transactions)


# Delete Discount column
del transactions["Discount"]

print("\nAfter Deleting Discount Column")
print(transactions)
```
