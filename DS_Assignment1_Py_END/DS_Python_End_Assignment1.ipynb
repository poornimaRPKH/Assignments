```python
#Task 3 – Create SaaS Customer Class (OOP)
# # Create Class called SaaSCustomer.
class SaaSCustomer:

    # Constructor Method
    def __init__(self, customer_id, name, signup_date, plan, monthly_fee, usage_hours):

        self.customer_id = customer_id
        self.name = name
        self.signup_date = signup_date
        self.plan = plan
        self.monthly_fee = monthly_fee
        self.usage_hours = usage_hours

    # Validate Data Method
    def validate_data(self):

        # Replace missing name
        if self.name == None or self.name == "":
            self.name = "Unknown Customer"

        # Standardize plan names
        self.plan = self.plan.title()

        # Convert monthly fee to float
        try:
            self.monthly_fee = float(self.monthly_fee)

        except:
            self.monthly_fee = 0.0

        # Convert usage hours to integer
        try:
            self.usage_hours = int(self.usage_hours)

        except:
            self.usage_hours = 0

    # Display Customer Method
    def display_customer(self):

        print("Customer Name:", self.name)


# Create Object
customer1 = SaaSCustomer(
    101,
    None,
    "2023-01-10",
    "pro",
    "49.99",
    "120"
)

# Call Methods
customer1.validate_data()
customer1.display_customer()

# Print Full Data
print(customer1.customer_id)
print(customer1.name)
print(customer1.signup_date)
print(customer1.plan)
print(customer1.monthly_fee)
print(customer1.usage_hours)
```

```python
#Task 4 – Data Cleaning Pipeline
from datetime import datetime

# Dataset
saas_data = {
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "name": ["Rahul Sharma", "Meena Nair", None, "A. Kumar", "David", "Priya", "Rahul Sharma"],
    "signup_date": ["2023-01-10", "2023/02/15", "03-01-2023", "15-04-2023", "2023-05-10", "20230520", "2023-01-10"],
    "plan": ["Pro", "Basic", "Pro", "Enterprise", "Pro", "", "Pro"],
    "monthly_fee": ["49.99", "19.99", "forty-nine", "99.99", "49.99", "49.99", "49.99"],
    "usage_hours": ["120", "45", "88", None, "NULL", "200", "120"]
}

# Clean Missing Values
for i in range(len(saas_data["customer_id"])):

    # Name
    if saas_data["name"][i] == None:
        saas_data["name"][i] = "Unknown Customer"

    # Usage Hours
    if saas_data["usage_hours"][i] == None or saas_data["usage_hours"][i] == "NULL":
        saas_data["usage_hours"][i] = 0

    # Monthly Fee
    try:
        saas_data["monthly_fee"][i] = float(saas_data["monthly_fee"][i])

    except:
        saas_data["monthly_fee"][i] = None

# Standardize Dates
for i in range(len(saas_data["signup_date"])):

    date = saas_data["signup_date"][i]

    try:
        new_date = datetime.strptime(date, "%Y/%m/%d")
        saas_data["signup_date"][i] = new_date.strftime("%Y-%m-%d")

    except:
        pass

# Print Cleaned Data
for i in range(len(saas_data["customer_id"])):

    print(
        saas_data["customer_id"][i],
        saas_data["name"][i],
        saas_data["signup_date"][i],
        saas_data["monthly_fee"][i],
        saas_data["usage_hours"][i]
    )
```

```python
#Task 5 – Summary Statistics and Insights
# Dataset
saas_data = {
    "customer_id": [101, 102, 103, 104, 105, 106],
    "name": ["Rahul Sharma", "Meena Nair", "Unknown Customer", "A. Kumar", "David", "Priya"],
    "plan": ["Pro", "Basic", "Pro", "Enterprise", "Pro", "Unknown"],
    "monthly_fee": [49.99, 19.99, None, 99.99, 49.99, 49.99],
    "usage_hours": [120, 45, 88, 0, 0, 200]
}

# Total Customers
total_customers = len(saas_data["customer_id"])

# Average Monthly Fee
total_fee = 0
count = 0

for fee in saas_data["monthly_fee"]:

    if fee != None:
        total_fee = total_fee + fee
        count = count + 1

average_fee = total_fee / count

# Total Usage Hours
total_usage = 0

for hours in saas_data["usage_hours"]:
    total_usage = total_usage + hours

# Plan Distribution
plan_count = {}

for plan in saas_data["plan"]:

    if plan in plan_count:
        plan_count[plan] = plan_count[plan] + 1

    else:
        plan_count[plan] = 1

# Print Results
print("Total Customers:", total_customers)

print("Average Monthly Fee:", round(average_fee, 2))

print("Total Usage Hours:", total_usage)

print("\nPlan Distribution")

for plan in plan_count:
    print(plan, "-", plan_count[plan])
```

```python
#Task 6 Identify Low-Usage Customers
# Dataset
saas_data = {
    "customer_id": [101, 102, 103, 104, 105],
    "name": ["Rahul Sharma", "Meena Nair", "Arun", "David", "Priya"],
    "plan": ["Pro", "Basic", "Pro", "Enterprise", "Basic"],
    "usage_hours": [120, 10, 88, 5, 200]
}

# Function
def find_low_usage_customers(data):

    print("Low Usage Customers\n")

    for i in range(len(data["customer_id"])):

        if data["usage_hours"][i] < 50:

            print(
                "ID:", data["customer_id"][i],
                "| Name:", data["name"][i],
                "| Plan:", data["plan"][i],
                "| Usage Hours:", data["usage_hours"][i]
            )

# Function Call
find_low_usage_customers(saas_data)
```
