```python
# Calculate the correlation matrix
correlation_matrix = mobile_data.corr(numeric_only=True)

# Set figure size
plt.figure(figsize=(12,8))

# Draw the heatmap parameters
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.5
)

# Add title
plt.title("Correlation Heatmap of Mobile Price Dataset")

# Display the heatmap
plt.show()
```

Split the Dataset

The dataset is divided into:
80% Training Data → Used to train the model.
20% Testing Data → Used to evaluate the model.
This helps us check how well the model performs on unseen data.

```python
# Import train_test_split
from sklearn.model_selection import train_test_split
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shapes
print("Training Feature Shape :", X_train.shape)
print("Testing Feature Shape  :", X_test.shape)

print("Training Target Shape  :", y_train.shape)
print("Testing Target Shape   :", y_test.shape)#
```

Create a Linear Regression Model

A Linear Regression model is one of the simplest Machine Learning algorithms. It predicts the value of a target variable (Price) based on the relationship between the input features and the target.

In this assignment, the model learns how features such as RAM, Battery, PPI, Camera, and Internal Memory affect the mobile price.

```python
# Import the Linear Regression model
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

print("Linear Regression model created successfully!")
```

```python
# Train the Linear Regression model
linear_model.fit(X_train, y_train)

print("Model training completed successfully!")
```

Assignment Task 4: Evaluate the Mode
Predict: Use the model to make predictions on the test set.

```python
# Predict mobile prices using the test data
predicted_price = linear_model.predict(X_test)

print("Predicted Prices:")
print(predicted_price)
```

Display the Slope (Coefficient) and Intercept
Code

```python
# Display the slope (coefficient) for each feature
coefficient_table = pd.DataFrame({
    "Feature": X.columns,
    "Slope (Coefficient)": linear_model.coef_
})

print("Slope (Coefficient) of Each Feature")
display(coefficient_table)

# Display the intercept
print("Intercept of the Regression Model:")
print(linear_model.intercept_)
```

Calculate the Model Performance Metrics
Code

```python
# Import evaluation metrics
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
# Calculate evaluation metrics
r2 = r2_score(y_test, predicted_price)
mae = mean_absolute_error(y_test, predicted_price)
mse = mean_squared_error(y_test, predicted_price)

# Display the results
print("Model Performance")
print("-" * 30)
print(f"R² Score               : {r2:.4f}")
print(f"Mean Absolute Error    : {mae:.2f}")
print(f"Mean Squared Error     : {mse:.2f}")
```

```python
# Display model evaluation metrics in a table

metrics_table = pd.DataFrame({
    "Metric": ["R² Score", "Mean Absolute Error (MAE)", "Mean Squared Error (MSE)"],
    "Value": [r2, mae, mse]
})

metrics_table
```
